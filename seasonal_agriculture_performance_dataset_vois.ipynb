```python
# STEP 1: REMOVE COMPLETE DUPLICATES
# ==========================================
# Removes any identical rows accidentally entered twice
duplicate_count = df.duplicated().sum()
if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print(f" Removed {duplicate_count} duplicate rows.")
```

```python
# STEP 2: STANDARDIZE TEXT (CATEGORICALS)
# ==========================================
# Strips hidden white spaces and enforces proper capitalization
categorical_cols = ['State', 'District', 'Crop', 'Season', 'Irrigation_Method']
for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.title()
print(" Standardized text formatting for categorical columns.")
```

```python
# STEP 3: IMPUTE MISSING VALUES WITH LOCAL MEDIANS
# ==========================================
# 1. Fill missing Rainfall using the median of that specific District and Season
if 'Rainfall_mm' in df.columns:
    df['Rainfall_mm'] = df.groupby(['District', 'Season'])['Rainfall_mm'].transform(
        lambda x: x.fillna(x.median())
    )
```

```python
# 2. Fill missing Soil Moisture using the median of that specific Irrigation Method
if 'Soil_Moisture_pct' in df.columns:
    df['Soil_Moisture_pct'] = df.groupby(['Irrigation_Method'])['Soil_Moisture_pct'].transform(
        lambda x: x.fillna(x.median())
    )
```

```python
# 3. Logically recalculate Yield if it's missing (Yield = Production / Area)
if 'yield_Tonnes_Ha' in df.columns:
    missing_yield_mask = df['yield_Tonnes_Ha'].isna()
    df.loc[missing_yield_mask, 'yield_Tonnes_Ha'] = (
        df.loc[missing_yield_mask, 'production_Tonnes'] /
        df.loc[missing_yield_mask, 'Farm_Area_Hectares']
    )

print(" Successfully handled missing values using domain grouping logic.")
```

```python
# STEP 4: ENFORCE REAL-WORLD BIOLOGICAL LOGIC
# ==========================================
# 1. Soil pH must strictly be between 0 and 14. Drop or cap anything outside.
if 'Soil_pH' in df.columns:
    df = df[(df['Soil_pH'] >= 0) & (df['Soil_pH'] <= 14)]

# 2. Percentages cannot exceed 100% or fall below 0%
pct_cols = ['Humidity_pct', 'Soil_Moisture_pct', 'Disease_Pest_Risk_pct']
for col in pct_cols:
    if col in df.columns:
        df = df[(df[col] >= 0) & (df[col] <= 100)]

print(" Cleaned biological limits (pH scales and percentage bounds verified).")
```

```python
# STEP 5: AUDIT & ENFORCE FINANCIAL LOGIC
# ==========================================
# Ensure that Profit always perfectly reflects: Revenue - Cost
if all(col in df.columns for col in ['profit_INR', 'Revenue_INR', 'Total_Cost_INR']):
    df['profit_INR'] = df['Revenue_INR'] - df['Total_Cost_INR']
print(" Recalculated financial columns to fix systemic bookkeeping errors.")
```

```python
# STEP 6: HANDLING OUTLIERS VIA CAPPING (WINSORIZATION)
# ==========================================
# We cap extreme values in continuous variables to minimize their skewing effect
continuous_numerical_cols = [
    'Farm_Area_Hectares', 'Rainfall_mm', 'Fertilizer_kg_ha',
    'production_Tonnes', 'yield_Tonnes_Ha', 'Revenue_INR', 'Total_Cost_INR'
]

for col in continuous_numerical_cols:
    if col in df.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Apply the caps (Winsorization technique)
        df[col] = np.where(df[col] < lower_bound, lower_bound, df[col])
        df[col] = np.where(df[col] > upper_bound, upper_bound, df[col])

print("✔️ Successfully capped extreme mathematical outliers using the IQR rule.")
print("------------------------------------------------------------------------")
print(f"Final Cleaned Dataset Shape: {df.shape}")

```

```python
# 2. Convert hidden whitespace string inputs into valid NumPy NaNs
df = df.replace(r'^\s*$', np.nan, regex=True)

```

```python
# 3. Print evaluation matrix tracking explicit missing gaps per column
print("--- Initial Gaps Detected Per Feature Variable ---")
print(df.isnull().sum())
```

```python
# 4. Handle Missing Values using Strategic Grouped Structural Imputation
# Rather than relying on inaccurate global averages, we fill missing elements with the
# median of fields matching the exact same Crop identity in that specific Season window.
numeric_cols_with_nan = [
    'Rainfall_mm', 'Avg_Temperature_C', 'Humidity_pct', 'Sunlight_Hours_Day',
    'Soil_pH', 'Soil_Moisture_pct', 'Nitrogen_kg_ha', 'Phosphorus_kg_ha',
    'Potassium_kg_ha', 'Fertilizer_kg_ha', 'Pesticide_Litre_ha', 'Yield_Tonnes_Ha'
]
```

```python
for col in numeric_cols_with_nan:
    df[col] = df.groupby(['Crop', 'Season'])[col].transform(lambda x: x.fillna(x.median()))
```

```python
print(df.isnull().sum())
```

```python
# 6. Formula Verification: Re-engineer mathematical and operational variables
# This rules out any corruption errors or human recording typos from the raw document source.
print("\n--- Re-Verifying Data Infrastructure Financial Equations ---")
df['Production_Tonnes'] = df['Farm_Area_Hectares'] * df['Yield_Tonnes_Ha']
df['Revenue_INR'] = df['Production_Tonnes'] * df['Market_Price_INR_Tonne']
df['Profit_INR'] = df['Revenue_INR'] - df['Total_Cost_INR']

print("--- Step 1 Pipeline Cleaned Execution Successfully Finalized ---")
print(df.dtypes)
print(f"Final Stable Dataset Structural Shape: {df.shape}")
```

```python
# 3. Statistical Group Contrast Validation Routine (One-Way ANOVA Framework)
yield_kharif = df[df['Season'] == 'Kharif']['Yield_Tonnes_Ha'].dropna()
yield_rabi = df[df['Season'] == 'Rabi']['Yield_Tonnes_Ha'].dropna()
yield_zaid = df[df['Season'] == 'Zaid']['Yield_Tonnes_Ha'].dropna()

print(f"\nGroup Cohort Verification Bounds -> Kharif: {len(yield_kharif)}, Rabi: {len(yield_rabi)}, Zaid: {len(yield_zaid)}")

f_stat, p_val = stats.f_oneway(yield_kharif, yield_rabi, yield_zaid)
print(f"\n--- One-Way ANOVA Statistical Validation Summary Matrix ---")
print(f"Computed F-Statistic Boundary Result Value: {f_stat:.4f}")
print(f"Computed Test Significance P-Value Scalar : {p_val:.4e}")

if p_val < 0.05:
    print("Statistical Verdict: Reject Null Hypothesis. Seasonal agricultural variance indicators are mathematically valid.")
else:
    print("Statistical Verdict: Accept Null Hypothesis. Variances can be classified as random distribution noise elements.")
```

```python
print("=== Phase 4: Production Allocation & Strategy Guidelines ===")

# 1. Strategic Crop Optimization Pivot Table Framework
seasonal_crop_profit = df.pivot_table(
    values='Profit_INR',
    index='Crop',
    columns='Season',
    aggfunc='mean'
).fillna(0)

print("[Optimization Blueprint] Average Net Returns generated per Variable Pair Scenario:")
print(seasonal_crop_profit.round(2))
```
